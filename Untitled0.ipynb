```python
import requests
import time
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import folium
from IPython.display import display


class StationData:
    def __init__(self):
        self._raw = {
            "station": [
                "Китай-город", "Белорусская", "Полянка", "Деловой центр",
                "Беговая", "Медведково", "Выхино", "Бульвар Дмитрия Донского",
                "Капотня", "ВДНХ", "Сходненская", "Коломенская",
                "Коньково", "ЦСКА", "Алтуфьево", "Фрунзенская",
                "Славянский бульвар"
            ],
            "passenger_flow_k": [
                116, 144, 35, 129,
                63, 101, 114, 103,
                0, 135, 90, 77,
                49, 73, 67, 60,
                98],
            "price_per_m2_k": [
                534, 410, 533, 441,
                348, 265, 245, 262,
                255, 310, 272, 311,
                297, 347, 247, 496,
                368],
            "rent_per_m2_k": [
                350, 220, 280, 250,
                110, 170, 95, 80,
                50, 160, 115, 130,
                120, 150, 85, 180,
                105],
            "population_density": [
                10417, 10897, 5629, 10897, 7767, 22903, 16120, 10680,
                3990, 5203, 18303, 12324, 21248, 7645, 17151, 10425, 16081
            ],
            "lat": [
                55.7538, 55.7766, 55.7368, 55.7490,
                55.7740, 55.8876, 55.7147, 55.5698,
                55.6374, 55.8210, 55.8512, 55.6780,
                55.6336, 55.7848, 55.8982, 55.7274,
                55.7285],
            "lon": [
                37.6343, 37.5846, 37.6191, 37.5394,
                37.5442, 37.6619, 37.8189, 37.5773,
                37.7988, 37.6394, 37.4410, 37.6646,
                37.5193, 37.5313, 37.5870, 37.5807,
                37.4745]
        }

    def load(self):
        return pd.DataFrame(self._raw)

class TwoGISClient:
    BASE_URL    = "https://catalog.api.2gis.com/3.0/items"
    GEOCODE_URL = "https://catalog.api.2gis.com/3.0/items/geocode"

    SEARCH_QUERIES = {
        "flower_shops":     "цветочный магазин",
        "supermarkets":     "супермаркет",
        "offices":          "бизнес-центр",
        "restaurants":      "ресторан",
        "hotels":           "отель",
        "registry_offices": "загс",
    }

    def __init__(self, api_key, radius=500):
        self.api_key = api_key
        self.radius  = radius

    def _search(self, lat, lon, query):
        params = {
            "q":         query,
            "point":     f"{lon},{lat}",
            "radius":    self.radius,
            "key":       self.api_key,
            "page_size": 10,
        }
        try:
            response = requests.get(self.BASE_URL, params=params, timeout=10)
            data     = response.json()
            code     = data.get("meta", {}).get("code")
            if code == 404:
                return 0
            if code != 200:
                print(f"Ошибка API: {data['meta']}")
                return 0
            return min(data.get("result", {}).get("total", 0), 10)
        except Exception as e:
            print(f"Ошибка запроса: {e}")
            return 0

    def get_coordinates(self, station_name):
        params = {
            "q":      f"Москва метро {station_name}",
            "key":    self.api_key,
            "fields": "items.point",
        }
        try:
            response = requests.get(self.GEOCODE_URL, params=params, timeout=10)
            data     = response.json()
            code     = data.get("meta", {}).get("code")
            if code != 200:
                return None
            items = data.get("result", {}).get("items", [])
            if not items:
                return None
            point = items[0].get("point", {})
            lat, lon = point.get("lat"), point.get("lon")
            return (lat, lon) if lat and lon else None
        except Exception as e:
            print(f"Геокодер: {e}")
            return None

    def enrich_coordinates(self, df):
        print("Получаем координаты...")
        for idx, row in df.iterrows():
            result = self.get_coordinates(row["station"])
            if result:
                df.at[idx, "lat"] = result[0]
                df.at[idx, "lon"] = result[1]
                print(f"  {row['station']}: {result[0]}, {result[1]}")
            else:
                print(f"  {row['station']}: используем координаты из данных")
            time.sleep(0.3)
        return df

    def enrich_poi(self, df):
        results = {key: [] for key in self.SEARCH_QUERIES}
        for _, row in df.iterrows():
            print(f"{row['station']}...", end=" ")
            for key, query in self.SEARCH_QUERIES.items():
                count = self._search(row["lat"], row["lon"], query)
                results[key].append(count)
                time.sleep(0.3)
        for key, values in results.items():
            df[key] = values
        return df


class ScoreCalculator:
    def __init__(self, weights=None):
        self.weights = weights or {
            "flow":        0.10,
            "density":     0.10,
            "price":       0.20,
            "demand":      0.35,
            "competition": 0.25,
        }

    @staticmethod
    def _normalize(series):
        mn, mx = series.min(), series.max()
        if mx == mn:
            return pd.Series([0.5] * len(series), index=series.index)
        return (series - mn) / (mx - mn)

    def calculate(self, df):
        df = df.copy()

        df["norm_flow"]     = self._normalize(df["passenger_flow_k"])
        df["norm_price"]    = self._normalize(df["price_per_m2_k"])
        df["norm_density"]  = self._normalize(df["population_density"])
        df["norm_rent_inv"] = 1 - self._normalize(df["rent_per_m2_k"])

        df["demand_index"] = (
            0.35 * self._normalize(df["offices"]) +
            0.35 * self._normalize(df["restaurants"]) +
            0.20 * self._normalize(df["hotels"]) +
            0.10 * self._normalize(df["registry_offices"])
        )

        df["competition_penalty"] = (
            0.70 * (1 - self._normalize(df["flower_shops"])) +
            0.30 * (1 - self._normalize(df["supermarkets"]))
        )

        total_w = sum(self.weights.values())
        df["base_score"] = (
            (self.weights["flow"]        / total_w) * df["norm_flow"] +
            (self.weights["density"]     / total_w) * df["norm_density"] +
            (self.weights["price"]       / total_w) * df["norm_price"] +
            (self.weights["demand"]      / total_w) * df["demand_index"] +
            (self.weights["competition"] / total_w) * df["competition_penalty"]
        )

        df["rent_multiplier"] = 0.5 + 0.5 * df["norm_rent_inv"]
        df["score_raw"]       = df["base_score"] * df["rent_multiplier"]
        df["score"]           = (df["score_raw"] * 10).round(2)
        df                    = df.sort_values("score", ascending=False).reset_index(drop=True)
        df["rank"]            = df.index + 1
        return df


class Visualizer:
    COLORS = {
        "top":    "#1A5276",
        "mid":    "#2E86C1",
        "light":  "#85C1E9",
        "pale":   "#D6EAF8",
        "bg":     "#F4F9FF",
        "text":   "#0A1628",
        "accent": "#1A3A5C",
    }

    def _bar_color(self, rank):
        if rank <= 3:   return self.COLORS["top"]
        elif rank <= 8: return self.COLORS["mid"]
        else:           return self.COLORS["light"]

    def plot_ranking(self, df):
        fig = plt.figure(figsize=(18, 9))
        fig.patch.set_facecolor(self.COLORS["bg"])

        ax1 = fig.add_subplot(1, 2, 1)
        ax1.set_facecolor(self.COLORS["bg"])

        bar_colors = [self._bar_color(r+1) for r in range(len(df))]
        bars = ax1.barh(
            df["station"][::-1], df["score"][::-1],
            color=bar_colors[::-1], edgecolor=self.COLORS["accent"],
            linewidth=0.5, height=0.7
        )
        for bar, score in zip(bars, df["score"][::-1]):
            ax1.text(bar.get_width() + 0.05, bar.get_y() + bar.get_height()/2,
                     f"{score:.1f}", va="center", ha="left",
                     fontsize=10, color=self.COLORS["text"], fontweight="bold")

        ax1.set_xlim(0, 11.5)
        ax1.set_xlabel("Скор (0–10)", fontsize=12, color=self.COLORS["text"])
        ax1.set_title("Рейтинг локаций", fontsize=14,
                      fontweight="bold", color=self.COLORS["accent"], pad=15)
        ax1.axvline(x=df["score"].mean(), color=self.COLORS["accent"],
                    linestyle="--", alpha=0.7, linewidth=1.5,
                    label=f"Среднее: {df['score'].mean():.1f}")
        ax1.legend(fontsize=10)
        ax1.tick_params(colors=self.COLORS["text"])
        for spine in ax1.spines.values():
            spine.set_color(self.COLORS["light"])

        categories = ["Трафик", "Платёжеспо-\nсобность", "Спрос",
                      "Нет конкурентов", "Доступность\nаренды", "Плотность\nнаселения"]
        col_map    = ["norm_flow", "norm_price", "demand_index",
                      "competition_penalty", "norm_rent_inv", "norm_density"]

        angles = np.linspace(0, 2*np.pi, len(categories), endpoint=False).tolist()
        angles += angles[:1]

        ax_radar = fig.add_subplot(1, 2, 2, polar=True)
        ax_radar.set_facecolor(self.COLORS["bg"])

        radar_colors = [self.COLORS["top"], self.COLORS["mid"], self.COLORS["light"]]
        for i, (_, row) in enumerate(df.head(3).iterrows()):
            values = [row[c] for c in col_map] + [row[col_map[0]]]
            ax_radar.plot(angles, values, "o-", linewidth=2,
                          color=radar_colors[i], label=row["station"], markersize=5)
            ax_radar.fill(angles, values, alpha=0.15, color=radar_colors[i])

        ax_radar.set_xticks(angles[:-1])
        ax_radar.set_xticklabels(categories, fontsize=10, color=self.COLORS["text"])
        ax_radar.set_ylim(0, 1)
        ax_radar.grid(color=self.COLORS["light"], linewidth=0.8)
        ax_radar.set_title("Профиль топ-3", fontsize=13, fontweight="bold",
                           color=self.COLORS["accent"], pad=20, y=1.08)
        ax_radar.legend(loc="upper right", bbox_to_anchor=(1.35, 1.15), fontsize=10)

        plt.suptitle("Анализ локаций для цветочного магазина | Москва",
                     fontsize=16, fontweight="bold", color=self.COLORS["text"], y=1.01)
        plt.tight_layout()
        plt.savefig("flower_shop_scoring.png", dpi=150,
                    bbox_inches="tight", facecolor=self.COLORS["bg"])
        plt.show()

    def plot_heatmap(self, df):
        metrics = {
            "Пассажиропоток":      "norm_flow",
            "Платёжеспособность":  "norm_price",
            "Плотность населения": "norm_density",
            "Спрос":               "demand_index",
            "Нет конкурентов":     "competition_penalty",
            "Доступность аренды":  "norm_rent_inv",
            "Итоговый скор":       "score_raw",
        }

        stations      = df["station"].tolist()
        metric_labels = list(metrics.keys())
        metric_cols   = list(metrics.values())

        data = df[metric_cols].copy()
        data["score_raw"] = (data["score_raw"] - data["score_raw"].min()) / \
                            (data["score_raw"].max() - data["score_raw"].min())
        matrix = data.values

        fig, ax = plt.subplots(figsize=(14, 10))
        fig.patch.set_facecolor(self.COLORS["bg"])
        ax.set_facecolor(self.COLORS["bg"])

        im = ax.imshow(matrix, aspect="auto", cmap="Blues", vmin=0, vmax=1)
        ax.set_xticks(range(len(metric_labels)))
        ax.set_xticklabels(metric_labels, rotation=30, ha="right",
                           fontsize=11, color=self.COLORS["text"])
        ax.set_yticks(range(len(stations)))
        ax.set_yticklabels(stations, fontsize=11, color=self.COLORS["text"])

        for i in range(len(stations)):
            for j in range(len(metric_labels)):
                val = matrix[i, j]
                ax.text(j, i, f"{val:.2f}", ha="center", va="center",
                        fontsize=9, fontweight="bold",
                        color="white" if val > 0.6 else self.COLORS["text"])

        cbar = plt.colorbar(im, ax=ax, shrink=0.8, pad=0.02)
        cbar.set_label("Нормализованное значение (0–1)",
                       fontsize=11, color=self.COLORS["text"])

        ax.set_title("Сравнение станций по всем показателям",
                     fontsize=15, fontweight="bold",
                     color=self.COLORS["accent"], pad=20)

        for i in range(len(stations) + 1):
            ax.axhline(i - 0.5, color=self.COLORS["light"], linewidth=0.5)
        for j in range(len(metric_labels) + 1):
            ax.axvline(j - 0.5, color=self.COLORS["light"], linewidth=0.5)

        for rank_idx in range(3):
            row_idx = stations.index(df.iloc[rank_idx]["station"])
            ax.add_patch(plt.Rectangle(
                (-0.5, row_idx - 0.5), len(metric_labels), 1,
                fill=False, edgecolor=self.COLORS["top"],
                linewidth=2.5, zorder=3
            ))

        plt.tight_layout()
        plt.savefig("flower_shop_heatmap.png", dpi=150,
                    bbox_inches="tight", facecolor=self.COLORS["bg"])
        plt.show()

    def plot_map(self, df):
        m = folium.Map(location=[55.75, 37.61], zoom_start=10,
                       tiles="OpenStreetMap")

        for idx in range(len(df)):
            row   = df.iloc[idx]
            score = float(row["score"])
            size  = 8 + score * 1.5

            if score >= 7:   color = self.COLORS["top"]
            elif score >= 5: color = self.COLORS["mid"]
            else:            color = "#aaaaaa"

            popup_html = f"""
            <div style="font-family: Arial; width: 210px;">
                <h4 style="color: #1A3A5C; margin:0">📍 {row['station']}</h4>
                <hr style="margin:5px 0">
                <b>Скор: {score}/10</b> (#{int(row['rank'])})<br>
                Трафик: {row['passenger_flow_k']} тыс./сут.<br>
                Цена м²: {row['price_per_m2_k']} тыс. руб.<br>
                Аренда: {row['rent_per_m2_k']} тыс. руб./м²/год<br>
                Цветочных рядом: {row['flower_shops']}<br>
                Офисов рядом: {row['offices']}
            </div>
            """

            folium.CircleMarker(
                location=[float(row["lat"]), float(row["lon"])],
                radius=float(size),
                color=color, fill=True,
                fill_color=color, fill_opacity=0.8,
                popup=folium.Popup(popup_html, max_width=250),
                tooltip=f"{row['station']}: {score}/10"
            ).add_to(m)

        legend_html = """
        <div style="position: fixed; bottom: 30px; left: 30px; z-index: 1000;
                    background: white; padding: 15px; border-radius: 10px;
                    border: 2px solid #85C1E9; font-family: Arial;">
            <b style="color: #1A3A5C"> Скор локации</b><br>
            <span style="color:#1A5276">●</span> 7–10 — Отлично<br>
            <span style="color:#2E86C1">●</span> 5–7  — Хорошо<br>
            <span style="color:#aaaaaa">●</span> 0–5  — Слабо
        </div>
        """
        m.get_root().html.add_child(folium.Element(legend_html))
        m.save("flower_shop_map.html")
        display(m)
        print("Карта сохранена: flower_shop_map.html")


API_KEY_2GIS = ""

data   = StationData()
client = TwoGISClient(api_key=API_KEY_2GIS)
calc   = ScoreCalculator(weights={
    "flow":        0.10,
    "density":     0.10,
    "price":       0.20,
    "demand":      0.35,
    "competition": 0.25,
})
viz = Visualizer()

df = data.load()
df = client.enrich_coordinates(df)
df = client.enrich_poi(df)
df = calc.calculate(df)

viz.plot_ranking(df)
viz.plot_heatmap(df)
viz.plot_map(df)

print("\n ИТОГОВЫЙ РЕЙТИНГ:")
print(df[["rank", "station", "score", "passenger_flow_k",
          "rent_per_m2_k", "flower_shops"]].to_string(index=False))
```
